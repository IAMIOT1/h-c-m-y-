1. Định nghĩa và cách nhận diện

Trùng khóa (Key/Primary Key Duplicates):

Bản chất: Trùng lặp ở cột định danh chính (ví dụ: cùng Product_ID), nhưng các cột thông tin khác lại khác nhau (ví dụ: khác giá bán Price_VND, khác thành phố Seller_City, hoặc khác số lượng tồn kho Stock).

Nguyên nhân: Sản phẩm có thể được cập nhật giá/thông tin qua nhiều thời kỳ, hoặc dữ liệu được gom từ nhiều chi nhánh/nguồn khác nhau nhưng cùng dùng chung một mã quản lý.

Cách kiểm tra: Dùng df.duplicated(subset=['Product_ID']) (chỉ xét riêng cột mã).

Phần 15: Tính toán các chỉ số thống kê

```python
# Tạo bản sao chứa các cột số tạm thời dùng pd.to_numeric(..., errors='coerce')
df_num_temp = pd.DataFrame()
num_cols = ['Price_VND', 'Rating', 'Purchase_Count', 'Stock', 'Weight']

for col in num_cols:
    df_num_temp[col] = pd.to_numeric(df[col], errors='coerce')

# Tính toán các chỉ số thống kê chi tiết
stats_results = []
for col in num_cols:
    s = df_num_temp[col].dropna()
    count = s.count()
    mean = s.mean()
    median = s.median()
    min_val = s.min()
    max_val = s.max()
    std = s.std()
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    stats_results.append({
        'Column': col,
        'Count': count,
        'Mean': mean,
        'Median': median,
        'Min': min_val,
        'Max': max_val,
        'Std': std,
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr
    })

df_stats_audit = pd.DataFrame(stats_results).set_index('Column')
print("--- BẢNG THỐNG KÊ MÔ TẢ CHI TIẾT ---")
display(df_stats_audit)
```

Phần 22: Chuẩn hóa cột Seller_City (Hà Nội, TP. Hồ Chí Minh,...)

```python
# Hàm chuẩn hóa tên thành phố
def standardize_city(city):
    if pd.isna(city):
        return city
    city_str = str(city).strip().lower()

    if city_str in ['hn', 'ha noi', 'hà nội', 'thu đô hà nội']:
        return 'Hà Nội'
    elif city_str in ['hcm', 'tp.hcm', 'tp hcm', 'hồ chí minh', 'tp. hồ chí minh', 'thành phố hồ chí minh']:
        return 'TP. Hồ Chí Minh'
    else:
        # Giữ nguyên hoặc xử lý các thành phố khác nếu có
        return str(city).strip().title()

# Áp dụng hàm chuẩn hóa lên cột Seller_City
df['Seller_City'] = df['Seller_City'].apply(standardize_city)
```

Phần 23: Chuẩn hóa cột Has_Image về 1 và 0

```python
# Chuyển đổi các giá trị True/False, 'Yes'/'No', 'Có'/'Không' thành 1 và 0
def standardize_has_image(val):
    if pd.isna(val):
        return np.nan # Hoặc có thể điền mặc định là 0 tùy nghiệp vụ

    val_str = str(val).strip().lower()
    if val_str in ['1', 'true', 'yes', 'có', 'co']:
        return 1
    elif val_str in ['0', 'false', 'no', 'không', 'khong']:
        return 0
    else:
        return np.nan # Xử lý các giá trị lạ thành NaN để xử lý ở bước điền thiếu sau

df['Has_Image'] = df['Has_Image'].apply(standardize_has_image)
```

Bài 6 Chuyển đổi kiểu dữ liệu iên tự viết hàm xử lý các ký hiệu như đ, VND, sao, lượt, %, kg, g và hậu tố

Phần 25 : Chuyển về số VND (Xử lý “miễn phí” và giá âm)

```python
import re
import pandas as pd
import numpy as np

def clean_price(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower()

    # 1. Xử lý trường hợp "miễn phí" hoặc "free" -> quy đổi về 0 VND
    if 'miễn phí' in val_str or 'free' in val_str:
        return 0.0

    # 2. Loại bỏ các ký tự chữ, đơn vị (đ, VND,...) chỉ giữ lại số và dấu chấm
    val_str = re.sub(r'[^\d.]', '', val_str)

    try:
        # 3. Phân biệt dấu chấm phân cách hàng nghìn (ví dụ: 350.000 có đúng 3 chữ số sau dấu chấm)
        if '.' in val_str and len(val_str.split('.')[-1]) == 3:
            val_str = val_str.replace('.', '')

        price = float(val_str)

        # 4. Quyết định xử lý giá âm: Theo nghiệp vụ, giá sản phẩm không thể âm -> gán về NaN để xử lý điền thiếu sau
        if price < 0:
            return np.nan

        return price
    except:
        return np.nan

# Áp dụng vào cột Price_VND của bản sao df_clean
df_clean['Price_VND'] = df_clean['Price_VND'].apply(clean_price)
```

Phần 26 : Tách phần số, chuyển về float

```python
def clean_rating(val):
    if pd.isna(val):
        return np.nan
    # Tìm chuỗi số thập phân đầu tiên xuất hiện trong ô (ví dụ: '4.8 sao' -> '4.8')
    match = re.search(r'[\d.]+', str(val))
    if match:
        try:
            return float(match.group())
        except:
            return np.nan
    return np.nan

df_clean['Rating'] = df_clean['Rating'].apply(clean_rating)
```

Phần 27 : Chuyển “1.2k” thành 1200 và “850 lượt” thành 850

```python
def clean_purchase_count(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower()

    # Xử lý trường hợp có hậu tố 'k' (ví dụ: 1.2k -> 1200)
    if 'k' in val_str:
        val_str_clean = val_str.replace('k', '')
        match = re.search(r'[\d.]+', val_str_clean)
        if match:
            try:
                return float(match.group()) * 1000
            except:
                return np.nan

    # Xử lý các định dạng thông thường có chữ "lượt" hoặc khoảng trắng
    match = re.search(r'[\d.]+', val_str)
    if match:
        try:
            return float(match.group())
        except:
            return np.nan
    return np.nan

df_clean['Purchase_Count'] = df_clean['Purchase_Count'].apply(clean_purchase_count)
```

Phần 28 : Quy đổi toàn bộ về gram

```python
def clean_weight(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower()

    match = re.search(r'[\d.]+', val_str)
    if not match:
        return np.nan
    try:
        num = float(match.group())
        # Nếu có đơn vị 'kg', nhân 1000 để quy đổi sang gram
        if 'kg' in val_str:
            return num * 1000
        else:
            # Mặc định là gram (g) hoặc số thuần túy
            return num
    except:
        return np.nan

df_clean['Weight'] = df_clean['Weight'].apply(clean_weight)
```

Phần 29 : Chuyển phần trăm về tỷ lệ 0–1

```python
def clean_discount(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip()

    match = re.search(r'[\d.]+', val_str)
    if not match:
        return np.nan
    try:
        num = float(match.group())
        # Nếu có dấu % hoặc giá trị lớn hơn 1 (ví dụ: 15 hoặc 15%), chuyển về dạng tỷ lệ thập phân (chia 100)
        if '%' in val_str or num > 1.0:
            return num / 100.0
        return num
    except:
        return np.nan

df_clean['Discount'] = df_clean['Discount'].apply(clean_discount)
```

Phép biến đổi log giúp nén các giá trị lớn lại, đưa phân phối về gần dạng phân phối chuẩn (Gaussian), giúp các mô hình học máy (như hồi quy tuyến tính, neural network) hội tụ tốt hơn và tránh bị chi phối bởi các giá trị ngoại lệ (outliers).Khi nào không dùng được:Không dùng được khi dữ liệu chứa số âm ($\le 0$): Hàm logarit không xác định với số âm hoặc bằng 0 (dù np.log1p đã khắc phục được trường hợp bằng 0, nhưng số âm vẫn ra lỗi NaN).Khi ý nghĩa nghiệp vụ cần giữ nguyên đơn vị gốc: Trong một số bài toán cần giải thích trực quan số tiền thực tế cho cấp quản lý (ví dụ: báo cáo doanh thu bằng triệu VND), việc dùng thang đo log sẽ khó diễn giải hơn.

Phần 53: So sánh StandardScaler và RobustScaler cho các biến số và nêu lý do chọn
So sánh:

StandardScaler: Dựa vào Trung bình (Mean) và Độ lệch chuẩn (Std). Phương pháp này nhạy cảm mạnh với ngoại lệ (outliers). Nếu dữ liệu có các điểm cực đoan (như tồn kho 9999 hoặc giá siêu cao), mean và std sẽ bị kéo lệch, làm bóp méo toàn bộ dữ liệu còn lại.

RobustScaler: Dựa vào Trung vị (Median) và Khoảng tứ phân vị (IQR). Do sử dụng các vị trí trung tâm ít bị ảnh hưởng bởi giá trị ngoại lệ, phương pháp này kháng nhiễu cực kỳ tốt trước các giá trị cực đoan.

Lý do chọn:

Trong bộ dữ liệu thương mại điện tử thường xuyên xuất hiện các sản phẩm có giá hoặc tồn kho cực trị (outliers), ta nên ưu tiên sử dụng RobustScaler để đảm bảo quá trình chuẩn hóa ổn định và không bị bóp méo bởi các giá trị bất thường.


```python
from sklearn.preprocessing import RobustScaler

# Áp dụng RobustScaler cho các cột số
num_cols_to_scale = ['Price_VND', 'Purchase_Count', 'Stock', 'Estimated_Revenue']
valid_scale_cols = [col for col in num_cols_to_scale if col in df_clean.columns]

scaler = RobustScaler()
df_clean[[f'{col}_robust_scaled' for col in valid_scale_cols]] = scaler.fit_transform(df_clean[valid_scale_cols])

print("Đã chuẩn hóa thành công các biến số bằng RobustScaler!")
```

Phần 58: Tính ma trận tương quan cho các biến số sau làm sạch

```python
import pandas as pd
import numpy as np

# Chọn các cột số (numeric columns) có trong df_clean sau khi làm sạch và tạo đặc trưng
numeric_df = df_clean.select_dtypes(include=[np.number])

# Tính ma trận tương quan Pearson
corr_matrix = numeric_df.corr()

print("--- MA TRẬN TƯƠNG QUAN ---")
display(corr_matrix.head(8))
```

Phần 59: Vẽ heatmap và tìm các cặp có $\vert r \vert$ lớn

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5, cbar=True)
plt.title('Ma trận tương quan (Heatmap) giữa các biến số')
plt.tight_layout()
plt.show()

# Tìm các cặp biến có hệ số tương quan tuyệt đối lớn (|r| > 0.7 và khác 1.0)
threshold = 0.7
high_corr_pairs = []

for i in range(len(corr_matrix.columns)):
    for j in range(i):
        val = corr_matrix.iloc[i, j]
        if abs(val) > threshold:
            high_corr_pairs.append((corr_matrix.columns[i], corr_matrix.columns[j], val))

print(f"--- CÁC CẶP BIẾN CÓ TƯƠNG QUAN CAO (|r| > {threshold}) ---")
for pair in high_corr_pairs:
    print(f"{pair[0]} & {pair[1]}: r = {pair[2]:.2f}")
```

Phần 69: Vì sao không nên xóa tất cả điểm nằm ngoài ngưỡng IQR?Phương pháp IQR ($Q1 - 1.5 \times IQR$ đến $Q3 + 1.5 \times IQR$) chỉ mang tính chất thống kê toán học để phát hiện các giá trị nằm ở đuôi phân phối.Nếu xóa thô bạo tất cả các điểm này, ta sẽ vô tình xóa mất các giá trị ngoại lệ hợp lệ (Valid Outliers) mang thông tin cực kỳ quan trọng (ví dụ: các sản phẩm thuộc phân khúc siêu cao cấp, hàng giới hạn giá trị lớn).Thay vì xóa, ta nên dùng phương pháp Clipping (cắt biên) hoặc gắn cờ (flagging) để vừa giữ được thông tin mẫu, vừa hạn chế tác động tiêu cực đến mô hình học máy.

Do không có cột thời gian (Updated_Time), ta không thể xác định phiên bản nào là mới nhất để ưu tiên giữ lại hay loại bỏ.

Giải pháp an toàn:

Giữ nguyên các bản ghi hoặc dùng phương pháp gắn cờ (Duplicate_ID_Flag) để đánh dấu các dòng xung đột nhằm phục vụ việc kiểm tra thủ công của bộ phận kiểm định dữ liệu.

Nếu bắt buộc phải đưa vào mô hình học máy theo quy tắc "mỗi sản phẩm một dòng", việc chọn dòng đầu tiên (keep='first') mang tính ngẫu nhiên và phải được nêu rõ trong phần hạn chế của báo cáo.

Phần 71: StandardScaler và RobustScaler khác nhau ra sao khi Stock có giá trị 9999?
StandardScaler: Dựa vào Trung bình (Mean) và Độ lệch chuẩn (Std). Khi xuất hiện giá trị cực đoan như 9999, Mean và Std sẽ bị kéo lệch mạnh về phía giá trị lớn này. Kết quả là các giá trị tồn kho thông thường (ví dụ: 10, 20, 50) sau khi chuẩn hóa sẽ bị ép sát lại với nhau và trở nên gần như bằng 0, làm mất đi khả năng phân biệt.

RobustScaler: Dựa vào Trung vị (Median) và Khoảng tứ phân vị (IQR). Các giá trị này ít bị ảnh hưởng bởi giá trị ngoại lệ. Do đó, giá trị 9999 sẽ được xem là một ngoại lệ đứng ở đuôi phân phối mà không làm bóp méo quá trình chuẩn hóa của toàn bộ các dữ liệu còn lại.

Phần 72: Có nên mã hóa Category bằng 1, 2, 3, 4, 5 không? Vì sao?
Tuyệt đối không nên (trừ trường hợp biến đó có tính thứ bậc rõ ràng như: Thấp, Trung bình, Cao).

Vì sao: Việc gán số nguyên ngẫu nhiên (ví dụ: Áo = 1, Giày = 2, Sách = 3) vô tình tạo ra quan hệ thứ tự giả trong không gian toán học của mô hình (mô hình sẽ hiểu lầm rằng Sách > Giày > Áo, hoặc Sách cách Giày đúng một khoảng đơn vị). Điều này dẫn đến sự sai lệch nghiêm trọng trong quá trình huấn luyện. Phải sử dụng One-Hot Encoding cho các biến phân loại danh nghĩa này.

Phần 73: Nếu xây dựng mô hình, tại sao phải chia train/test trước khi fit imputer và scaler?
Để tránh hiện tượng rò rỉ dữ liệu (Data Leakage).

Nếu ta tính toán giá trị điền thiếu (Imputer như Median) hoặc tham số chuẩn hóa (Scaler như Mean/Std hoặc Median/IQR) trên toàn bộ tập dữ liệu trước khi chia, thông tin từ tập kiểm tra (Test set) đã vô tình bị "rò rỉ" sang tập huấn luyện (Train set). Điều này làm cho đánh giá độ chính xác của mô hình trên tập test trở nên quá lạc quan (overoptimistic) và thất bại khi đưa ra dự đoán thực tế.

Phần 74: Bộ dữ liệu sạch có phải là bộ dữ liệu không còn giá trị thiếu trong mọi trường hợp không?
Không phải lúc nào cũng vậy.

"Sạch" trong tiền xử lý dữ liệu đồng nghĩa với việc các giá trị thiếu đã được xử lý có chủ đích và phù hợp với nghiệp vụ, chứ không có nghĩa là xóa sạch mọi giá trị thiếu bằng mọi giá.

Ví dụ: Cột Created_Date bị thiếu được giữ nguyên dạng NaT vì nguyên tắc không được bịa ngày tháng; hoặc một số biến đặc thù được giữ lại giá trị thiếu và đi kèm một cột cờ (_Missing_Flag) để mô hình máy học tự học hỏi ý nghĩa từ việc thiếu dữ liệu đó.
