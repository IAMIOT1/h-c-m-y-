Phần 9: Phân biệt chi tiết Trùng hoàn toàn và Trùng khóa

1. Định nghĩa và cách nhận diện
Trùng hoàn toàn (Full Row Duplicates):

Bản chất: Các dòng có giá trị hoàn toàn giống nhau ở tất cả các cột (từ Product_ID, tên, giá, thành phố cho đến ngày tháng).

Cách kiểm tra: Dùng df.duplicated() (xét toàn bộ cột).

Trùng khóa (Key/Primary Key Duplicates):

Bản chất: Trùng lặp ở cột định danh chính (ví dụ: cùng Product_ID), nhưng các cột thông tin khác lại khác nhau (ví dụ: khác giá bán Price_VND, khác thành phố Seller_City, hoặc khác số lượng tồn kho Stock).

Nguyên nhân: Sản phẩm có thể được cập nhật giá/thông tin qua nhiều thời kỳ, hoặc dữ liệu được gom từ nhiều chi nhánh/nguồn khác nhau nhưng cùng dùng chung một mã quản lý.

Cách kiểm tra: Dùng df.duplicated(subset=['Product_ID']) (chỉ xét riêng cột mã).

```python
# --- BƯỚC 1: Quan sát dòng trùng hoàn toàn ---
du_thuan_hoan_toan = df[df.duplicated(keep=False)]
print(f"Số lượng dòng thuộc nhóm trùng hoàn toàn: {df.duplicated().sum()}")
if not du_thuan_hoan_toan.empty:
    print("Ví dụ các dòng trùng hoàn toàn (tất cả các cột y hệt nhau):")
    display(du_thuan_hoan_toan.head())

print("-" * 50)

# --- BƯỚC 2: Quan sát dòng trùng khóa (Product_ID trùng nhưng khác nội dung) ---
# Lọc các dòng có Product_ID bị lặp
du_khoa_product_id = df[df.duplicated(subset=['Product_ID'], keep=False)]

# Loại bỏ các dòng đã trùng hoàn toàn để chỉ tập trung vào nhóm "Cùng mã nhưng khác nội dung"
du_khoa_thuan_tu = du_khoa_product_id[~du_khoa_product_id.duplicated(keep=False)]

print(f"Số lượng bản ghi có Product_ID bị trùng lặp: {len(du_khoa_product_id)}")
if not du_khoa_product_id.empty:
    print("Danh sách các dòng có Product_ID bị trùng (Sắp xếp theo mã để so sánh giá/thành phố):")
    display(du_khoa_product_id.sort_values(by='Product_ID'))
```

Phần 15: Tính toán các chỉ số thống kê

```python
# Tạo bản sao chứa các cột số tạm thời dùng pd.to_numeric(..., errors='coerce')
df_num_temp = pd.DataFrame()
num_cols = ['Price_VND', 'Rating', 'Purchase_Count', 'Stock', 'Weight']

for col in num_cols:
    df_num_temp[col] = pd.to_numeric(df[col], errors='coerce')

# Tính toán các chỉ số thống kê chi tiết
stats_results = []
for col in num_cols:
    s = df_num_temp[col].dropna()
    count = s.count()
    mean = s.mean()
    median = s.median()
    min_val = s.min()
    max_val = s.max()
    std = s.std()
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    stats_results.append({
        'Column': col,
        'Count': count,
        'Mean': mean,
        'Median': median,
        'Min': min_val,
        'Max': max_val,
        'Std': std,
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr
    })

df_stats_audit = pd.DataFrame(stats_results).set_index('Column')
print("--- BẢNG THỐNG KÊ MÔ TẢ CHI TIẾT ---")
display(df_stats_audit)
```

Phần 22: Chuẩn hóa cột Seller_City (Hà Nội, TP. Hồ Chí Minh,...)

```python
# Hàm chuẩn hóa tên thành phố
def standardize_city(city):
    if pd.isna(city):
        return city
    city_str = str(city).strip().lower()

    if city_str in ['hn', 'ha noi', 'hà nội', 'thu đô hà nội']:
        return 'Hà Nội'
    elif city_str in ['hcm', 'tp.hcm', 'tp hcm', 'hồ chí minh', 'tp. hồ chí minh', 'thành phố hồ chí minh']:
        return 'TP. Hồ Chí Minh'
    else:
        # Giữ nguyên hoặc xử lý các thành phố khác nếu có
        return str(city).strip().title()

# Áp dụng hàm chuẩn hóa lên cột Seller_City
df['Seller_City'] = df['Seller_City'].apply(standardize_city)
```

Phần 23: Chuẩn hóa cột Has_Image về 1 và 0

```python
# Chuyển đổi các giá trị True/False, 'Yes'/'No', 'Có'/'Không' thành 1 và 0
def standardize_has_image(val):
    if pd.isna(val):
        return np.nan # Hoặc có thể điền mặc định là 0 tùy nghiệp vụ

    val_str = str(val).strip().lower()
    if val_str in ['1', 'true', 'yes', 'có', 'co']:
        return 1
    elif val_str in ['0', 'false', 'no', 'không', 'khong']:
        return 0
    else:
        return np.nan # Xử lý các giá trị lạ thành NaN để xử lý ở bước điền thiếu sau

df['Has_Image'] = df['Has_Image'].apply(standardize_has_image)
```
